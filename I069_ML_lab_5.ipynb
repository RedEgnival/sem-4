# Non vectorized Gradient Descent

```python
x = np.array([1,2,3,3,4,4,5])
y = np.array([80,170,100,220,200,270,500])
xy = np.sum(x*y)
x2 = np.sum(x**2)
xs = np.sum(x)
ys = np.sum(y)
n = len(x)
```

```python
iter = int(input("Enter np. of iterations "))
a = float(input("Enter learning rate "))
b0 = 0
b1 = 0
for i in range (iter):
  b0 = b0- a/n * np.sum(b0 + b1*x - y)
  b1 = b1 - a/n * np.sum((b0 + b1*x - y)*x)
  j = np.sum((b0 + b1*x - y)**2)/n
print("Cost function =",j)
print("Intercept and slope =",b0,b1)
x1 = int(input("Enter value to predict "))
print("Predicted y is",b0 + b1*x1)
```

```python
plt.scatter(x, y, color='blue', label='Data Points')  # Original data points
plt.plot(x,b0 + b1*x , color='red', label='Fitted Line')  # Fitted line
plt.xlabel('Number of Bedrooms')
plt.ylabel('Price')
plt.title('Linear Regression Fit')
plt.legend()
plt.grid()
plt.show()
```

# Vectorized Gradient Descent

```python
x1=np.array([1,2,3,3,4,4,5])
y1=np.array([80,170,100,220,200,270,500])
```

```python
itr = int(input("Enter no. of iterations: "))
a = float(input("Enter learning rate (alpha value): "))
t0 = 0
t1 = 0
for _ in range(itr):
    y_pred = t0 + t1 * x
    Grad0 = np.sum(y_pred - y)
    Grad1 = np.sum((y_pred - y) * x)
    t0 = t0 - a * (1/n) * Grad0
    t1 = t1 - a * (1/n) * Grad1
y_pred = t0 + t1 * x
cost = (1/(2*n)) * np.sum((y_pred - y) ** 2)

print("Theta0:", t0, "Theta1:", t1)
print("Cost J:", cost)
```

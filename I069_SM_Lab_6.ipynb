Q1. Create a pie chart, bar chart, and Pareto chart to represent the frequency of the following categories of issues reported in a helpdesk:

•	Login Issues: 40

•	Password Reset: 25

•	Network Issues: 15

•	Software Bugs: 20


```python
cat = [" Login Issues","Password Reset","Network Issues","Software Bugs"]
freq = [40,25,15,20]
```

Pie Chart

```python
plt.pie(freq,labels=cat, autopct = "%1.1f%%")

plt.title("Pie Chart")
```

1.	What percentage of the total does the largest category represent?

    40%
2.	If a new category is introduced, how would you adjust the pie chart?
    
    Redistributing proportions among the categories
3.	Does the pie chart effectively convey the differences between categories? Why or why not?
    
    Yes

Bar Chart

```python
plt.bar(cat,freq)

plt.title("Bar Chart")
plt.xlabel("Category")
plt.ylabel("Frequency")
```

Bar Chart
1.	Which category has the highest/lowest frequency?

  Highest Category: Login Issues

  Lowest Category: Network Issues
2.	If two categories had the same frequency, how would that appear in the bar chart?

    Bars would be of equal height
3.	How can you modify the bar chart to show additional information, like percentages?

    Add percentage Labels

Pareto Chart

```python
sorted_indices = np.argsort(freq)[::-1]
sorted_freq = [freq[i] for i in sorted_indices]
sorted_cat = [cat[i] for i in sorted_indices]
cum = np.cumsum(sorted_freq) / sum(freq)*100
plt.bar(sorted_cat, sorted_freq, label = "frequency")

plt.plot(sorted_cat, cum, color = "red",marker="o", label = "cumulative %")
plt.legend()
plt.title("Pareto Chart")
plt.xlabel("Categories")
plt.ylabel("Frequency/Cumulative %")
```

1.	What is the 80/20 rule, and how can you identify it from the Pareto chart?

    80% of problems come from 20% of the causes. Identify it by looking at the steepest rise in the cumulative frequency curve.
2.	Which categories contribute the most to the cumulative frequency?

    Login Issues and Password Reset contribute the most.
3.	How does the cumulative percentage line help in prioritizing categories?

     It shows the most impactful categories to focus on first.
4.	If you wanted to focus on solving 80% of the problems, which categories should you address first?

    Focus on Login Issues and Password Reset first.

Q2. Create a histogram, frequency polygon, ogive, dot plot, and stem-and-leaf plot for the following data representing the marks (out of 50) scored by 20 students:

[12, 25, 32, 18, 27, 29, 35, 41, 15, 19, 24, 37, 31, 42, 26, 17, 30, 28, 34, 40]

….all in one



```python
data = [12, 25, 32, 18, 27, 29, 35, 41, 15, 19, 24, 37, 31, 42, 26, 17, 30, 28, 34, 40]
```

Histogram

```python
plt.figure(figsize=(12,8))
plt.subplot(3,2,1)

plt.hist(data,bins = range(10,51,5), edgecolor = "black")
plt.title("Histogram")
plt.xlabel("Marks")
plt.ylabel("Frequency")
```

Histogram
1.	How many bins are in your histogram, and what is the bin width?

    7,5
2.	Which bin has the highest frequency? What does this tell you about the data?

    Bin 4
3.	Does the histogram suggest any patterns, such as a normal distribution?

    The histogram does not appear perfectly normal; it may have skewness or multiple peaks.


Frequency Polygon

```python
freq, bin_edges = np.histogram(data, bins = range(10,51,5))
midpoints = (bin_edges[:-1] + bin_edges[1:])/2
plt.plot(midpoints,freq,marker = "o", label="Frequency Polygon")
plt.title("Frequency Polygon")
plt.xlabel("Marks")
plt.ylabel("Frequency")
plt.grid()
```

Frequency Polygon
1.	How does the frequency polygon differ from a histogram in terms of visualization?

    A frequency polygon uses a continuous line connecting midpoints of bins, while a histogram uses bars.
2.	Does the frequency polygon make it easier to identify trends or peaks? Why or why not?

    Yes, it helps by clearly showing peaks and trends through connected points, making comparisons easier.


Ogive

```python
cum_f = np.cumsum(freq)
```

```python
plt.plot(bin_edges[:-1], np.insert(cum_f,0,0)[:-1], marker = "o", label = "Ogive")
plt.title("Ogive")
plt.xlabel("Marks")
plt.ylabel("Cumulative Frequency")
plt.grid()
```

Ogive
1.	Based on the ogive, how many students scored below 30?

    10
2.	What percentage of students scored above 35?

    25%
3.	Can you determine the median score from the ogive? How?

    Yes, Divide cumulative frequency in half

Dot Plot

```python
for i, mark in enumerate(sorted(data)):
  plt.scatter([mark],[i+1], s= 20)
plt.title("Dot Plot")
plt.xlabel("Marks")
plt.ylabel("Frequency")

```

Dot Plot Questions
1.	How many unique values are present in your dataset? Can you see this clearly on the dot plot?

    20, Yes
2.	Is there any value with a very high frequency? How would this look in a dot plot?

    A value with very high frequency would appear as a tall stack of dots at a single x-axis position. If no tall stacks exist, frequencies are more evenly distributed.

Steam and leaf Plot

```python
df = pd.DataFrame({"Marks" : data})
df["Stem"] = df["Marks"] // 10
df["Leaf"] = df["Marks"] % 10
stem_leaf = df.groupby("Stem")["Leaf"].apply(lambda x: sorted(x)).reset_index()

for _, row in stem_leaf.iterrows():
  stem = row["Stem"]
  leaves = "".join(map(str,row["Leaf"]))
  plt.text(0.1, 1-0.1 * stem, f"{stem} | {leaves}", fontsize = 10)
  plt.axis("off")
plt.title("Stem and Leaf Plot")
plt.tight_layout()
plt.show()
```

1.What does each row in the stem-and-leaf plot represent?

  Each row represents a group of values with the same leading digit and trailing digits show individual values.
2.How would you reconstruct the original dataset from the plot?

  Reconstruct data by combining each stem with its leaves to list original numbers.

3.Can you identify the highest and lowest scores? How?

  Highest & lowest scores are the largest and smallest leaves in the last and first rows, respectively.

4.What can you infer about the range and spread of the data from this plot?

  Range & spread are determined by the highest-lowest difference and leaf distribution.
5.How is the stem-and-leaf plot similar to a histogram? How is it different?

  Similar to histogram but different as it retains exact values while a histogram groups data.

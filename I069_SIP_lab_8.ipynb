```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import signal
import cv2
```

# Edge Detection

```python
img = cv2.imread('/content/house1.jpg',0)
plt.imshow(img, cmap = "grey", vmin = 0, vmax = 255)
m,n = img.shape
```

```python
img.shape
```

# Using in-build functions

Detect horizontal edges

```python
sobel_h = np.array([[-1,-2,-1],[0,0,0],[1,2,1]])
imgh = signal.convolve(img, sobel_h, mode = 'same')
plt.imshow(imgh, cmap = "grey", vmin = 0, vmax = 255)
```

Verticl edges

```python
sobel_v = np.array([[-1,0,1],[-2,0,2],[-1,0,1]])
imgv = signal.convolve(img, sobel_v, mode = 'same')
plt.imshow(imgv, cmap = "grey", vmin = 0, vmax = 255)
```

```python
imghv = imgh + imgv
plt.imshow(imghv, cmap = "grey", vmin = 0, vmax = 255)
```

Diagonal edges



```python
sobel_d = np.array([[0,1,2],[-1,0,1],[-2,-1,0]])
imgd = signal.convolve(img, sobel_d, mode = 'same')
plt.imshow(imgd, cmap = "grey", vmin = 0, vmax = 255)
```

```python
imghvd = imgh + imgv + imgd
plt.imshow(imghvd, cmap = "grey", vmin = 0, vmax = 255)
```

# Without using inbuilt functions

Horizontal edges

```python
img_h = np.zeros([m,n])
mask = np.array([[-1,-2,-1],[0,0,0],[1,2,1]])
for i in range (1,m-1):
  for j in range (1,n-1):
    temp = img[i-1,j-1]*mask[0,0] + img[i-1,j]*mask[0,1] + img[i-1,j+1]*mask[0,2] + img[i,j-1]*mask[1,0]
    + img[i,j]*mask[1,1] + img[i,j+1]*mask[1,2] + img[i+1,j-1]*mask[2,0] + img[i+1,j]*mask[2,1] + img[i+1,j+1]*mask[2,2]
    new_h[i,j] = temp
plt.imshow(new_h, cmap = "grey", vmin = 0, vmax = 255)
```

Vertical edges

```python
mask_v = np.array([[-1,0,1],[-2,0,2],[-1,0,1]])
img_v = np.zeros([m,n])
for i in range (1,m-1):
  for j in range (1,n-1):
    temp = img[i-1,j-1]*mask_v[0,0] + img[i-1,j]*mask_v[0,1] + img[i-1,j+1]*mask_v[0,2] + img[i,j-1]*mask_v[1,0]+ img[i,j]*mask_v[1,1] + img[i,j+1]*mask_v[1,2]
    + img[i+1,j-1]*mask_v[2,0] + img[i+1,j]*mask_v[2,1] + img[i+1,j+1]*mask_v[2,2]
    img_v[i,j] = temp
plt.imshow(img_v, cmap = "grey", vmin = 0, vmax = 255)
```

Diagonal edges

```python
mask_d = np.array([[0,1,2],[-1,0,1],[-2,-1,0]])
img_d = np.zeros([m,n])
for i in range(1,m-1):
  for j in range(1,n-1):
    temp = img[i-1,j-1]*mask_d[0,0] + img[i-1,j]*mask_d[0,1] + img[i-1,j+1]*mask_d[0,2] + img[i,j-1]*mask_d[1,0]+ img[i,j]*mask_d[1,1] + img[i,j+1]*mask_d[1,2]
    + img[i+1,j-1]*mask_d[2,0] + img[i+1,j]*mask_d[2,1] + img[i+1,j+1]*mask_d[2,2]
    img_d[i,j] = temp
plt.imshow(img_d, cmap = "grey", vmin = 0, vmax = 255)
```

```python
img_hvd = img_h + img_v + img_d
plt.imshow(img_hvd, cmap = "grey", vmin = 0, vmax = 255)
```

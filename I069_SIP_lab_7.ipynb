```python
N=4
W_inv=np.zeros((4,4), dtype=complex)
for n in range(0,N):
  for k in range(0,N):
    real = int(np.cos(2*np.pi*n*k/N))
    imaginary = int(np.sin(2*np.pi*n*k/N))
    W_inv[n,k]=complex(real,imaginary)
print(W_inv)
```

```python
N=8
W=np.zeros((8,8),dtype=complex)
for n in range(0,N):
  for k in range(0,N):
    real = np.cos(2*np.pi*n*k/N)
    imaginary = -1*(np.sin(2*np.pi*n*k/N))
    W[n,k]=np.round(complex(real,imaginary))
print(W)
```

```python
x2=np.array([1,1,1,1,1,1,0,0])
```

```python
X2=np.dot(W,np.transpose(x2))
for i in range(0,7):
  print(X2[i])

X2new=np.fft.fft(x2, axis=-1, norm=None)
for i in range(0,7):
  print(X2new[i])
```
